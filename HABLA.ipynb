```python
import os
```

```python
HABLA = './HABLA/data/'
```

```python
tts = []
reales = []
count = 0

for file in os.listdir(HABLA):
    count += 1
    file = file[:-4]
    if file.endswith('TTS'): tts.append(file)
    
for file in os.listdir(HABLA):
    file = file[:-4]
    if file + '_TTS' in tts: reales.append(file)
```

```python
count
```

```python
print(f'Len tts: {len(tts)}')
print(f'Len pares: {len(reales)}')
```

```python
pares = []
for real in reales:
    for generado in tts:
        if real == generado[:-4]:
            pares.append((real, generado))
            break
```

```python
print(len(pares))
pares
```
